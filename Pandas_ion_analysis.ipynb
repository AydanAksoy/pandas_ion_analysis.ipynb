```python
import pandas as pd
import numpy as np

# pandas ist wie Excel in Python
# pandas — это как Excel в Python

# eine Tabelle erstellen / создаём таблицу
data = {
    'Ion':        ['H', 'He', 'C',  'U'],
    'Masse':      [1,    4,   12,   238],
    'Energie_MeV':[1.0,  2.0, 10.0, 50.0],
    'Reichweite': [15.2, 8.4, 3.1,  0.8]
}

df = pd.DataFrame(data)

print("Meine Ionen-Tabelle / Моя таблица ионов:")
print(df)
print()
print("Schwerste Ion / Самый тяжёлый ион:")
print(df[df['Masse'] == df['Masse'].max()])
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Uranium bei verschiedenen Energien
# Уран при разных энергиях
data = {
    'Energie_MeV': [10, 20, 50, 100, 200, 500],
    'Reichweite_mm': [0.2, 0.4, 0.8, 1.5, 3.2, 8.1]
}

df = pd.DataFrame(data)

print(df)
print()
print("Mittelwert Reichweite / Среднее пробега:", 
      round(df['Reichweite_mm'].mean(), 2), "mm")
print("Max Reichweite / Макс пробег:", 
      df['Reichweite_mm'].max(), "mm")

# Graph / График
plt.figure(figsize=(8, 5))
plt.plot(df['Energie_MeV'], df['Reichweite_mm'], 
         'o-', color='steelblue', linewidth=2, markersize=8)
plt.xlabel('Energie / МэВ')
plt.ylabel('Reichweite / Пробег (mm)')
plt.title('Uranium Ionen — Energie vs Reichweite\n'
          'Уран — Энергия vs Пробег')
plt.grid(True, alpha=0.3)
plt.show()
```
